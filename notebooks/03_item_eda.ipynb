```python
import pandas as pd
```

```python
df = pd.read_csv('../data/raw/on_items.csv', encoding='cp949')
df
```

이상치 제거

```python
# 스테인리스: L1-M11-S0080-1024
df = df[df['ItemCode'] != 'L1-M11-S0080-1024']
```

이상치 대체

```python
# 전복죽 소분류명 ->S0540인 전복을 즉석으로 대체
df.loc[df['ItemSmallCode'] == 'S0540', 'ItemSmallName'] = '즉석'

# S0240인 소분류명을 마른안주로 대체
df.loc[df['ItemSmallCode'] == 'S0240', 'ItemSmallName'] = '마른안주'

# S0720인 소분류명을 혼합및이온음료로 대체
df.loc[df['ItemSmallCode'] == 'S0720', 'ItemSmallName'] = '혼합및이온음료'
```

```python
print("스테인리스 제거 여부:", 'L1-M11-S0080-1024' in df['ItemCode'].values)

# S0240 소분류명 확인
print(df[df['ItemSmallCode'] == 'S0240'])
```

```python
# S0720 소분류명 확인
print(df[df['ItemSmallCode'] == 'S0720'])
```
